```python
# Data Types
print("--------------------------------")

# Integer
print(type(8))

# float/Decimal/double
print(type(8.55))

# complex
print(type(3+5j))

# Boolean
print(type(True))

# string ---> sequence of character
print(type("Shiv mohan"))

# List
print(type([12,"hi",True]))

# Tuple
print(type((12,"hi",True)))

# set
print(type({12,"hi",True}))

# dictionary
info={
    "name":"prince",
    "gender":"male",
    "age":30
}
print(type(info))

# None --->variable decleare for useful in future 
a=None
print(a)
print(type(a))
```
