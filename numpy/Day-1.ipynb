```python
# numpy --> python library that handle eaisly large numsber of data using array , it is fast and less memmory used

# create array from python list

import numpy as np

arr=np.array([10,20,30,40,50,60])
print(arr)
print(type(arr))
```

```python
arr_zeros=np.zeros((5,3))
print(arr_zeros)
```

```python
arr_ones=np.ones((2,3))
print(arr_ones)
```

```python
arr=np.full((2,3),6)
print(arr)
```

```python
arr=np.arange(1,10,2)
print(arr)
```

```python
# create identity matrix

arr=np.eye(4)
print(arr)
```

```python
temp=[[1,2,3],[4,5,6]]
arr=np.array(temp)
print(arr)

print("print no. of rows and columns of array ")
print(arr.shape)

print("print total element in array")
print(arr.size)

print("no. of dimension of array")
print(arr.ndim)

print("Find data type of element")
print(arr.dtype)

print("convert on data type into other")
arr2=arr.astype(float)
print(arr2)
print(arr2.dtype)
```

```python
# mathemaical operation  ->  + - * /  **  //  %


arr=np.array([1,2,3,4,5,6])
print(arr+3)
print(arr*2)
print(arr**2)
```

```python
# Aggregation funtion   --> sum , mean , max, min, std, var


arr=np.array([1,2,3,4,5,6])
print(np.sum(arr))
print(np.mean(arr))
print(np.min(arr))
print(np.max(arr))
print(np.std(arr))
print(np.var(arr))
```

```python
# indexing  ---> positive and negative indeing
# 1-D array  --> arr[index]
# 2-D array  --> arr[row,column]
arr=np.array([[1,2,3],
              [4,5,6]])
print(arr)
print(arr[0,1])
print(arr[1,1])
```

```python
# slicing ---> arr[start : stop : step]

arr=np.array([10,20,30,40,50,60])
print(arr[2: : 2])
print(arr[::-1])
```

```python
# fancy indexing --> selecting multiple elements at onceby using index
arr=np.array([10,20,30,40,50,60])
print(arr[[0,3,5]])
```

```python
# Boolean masking  ---> filtering data based on condition
arr=np.array([10,20,30,40,50,60])
print(arr[arr>25])
```

```python
# reshaping of array dimension ----> convert lower dimension array into higher
# arr.reshape(row , column)  ---> it return a view (change in original array)
arr=np.array([10,20,30,40,50,60])
print(arr.reshape(2,3))
print(arr)
```

```python
# flatening array ---> convert higher dimension array into 1-D array 
# arr.ravel()
# arr.flatten()
arr=np.array([[10,20,30],[40,50,60]])
print(arr.ravel())
```
