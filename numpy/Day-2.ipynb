```python
# modification on new array from existing array
# np.insert(arr,index,value,axis=None)

import numpy as np
arr=np.array([10,20,30,40])
print(arr)

# 1-D array
new_arr=np.insert(arr,1,1000,axis=None)
print(new_arr)

# 2-array
arr1=np.array([[1,2,3],[4,5,6]])
print(arr1)
arr2=np.insert(arr1,1,[10,20,30],axis=0)
print(arr2)
```

```python
# np.append(arr,list)

arr=np.array([10,20,30,40])
new_arr=np.append(arr,[10,20,30])
print(new_arr)
```

```python
# np.concatenate ((arr1,arr2,......),axis=0)

arr1=np.array([10,20,30,40])
arr2=np.array([50,60,30,70])

ans=np.concatenate((arr1,arr2),axis=0)
print(ans)

ans3=np.concatenate((arr1,arr2),axis=0)
print(ans3)
```
