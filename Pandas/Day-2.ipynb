```python
# detect missing value
print(df.isnull().sum())

data2 ={
    "Name":["Ajay","Raj","Shiva"],
    "Age":[20,None,22],
    "Salary":[5000,3000,4000]
}

df2=pd.DataFrame(data2)
print(df2.isnull().sum())
```

```python
# Remove missing value
df2.dropna(inplace=True)
df2
```

```python
# Fill with custom value in missing place
data3 ={
    "Name":["Ajay","Raj",None],
    "Age":[20,None,22],
    "Salary":[5000,3000,4000]
}
df3=pd.DataFrame(data3)
df3.fillna(0,inplace=True)
df3
```

```python
# fill with estimate value

data4 ={
    "Name":["Ajay","Raj",None],
    "Age":[20,None,22],
    "Salary":[5000,3000,4000]
}
df4=pd.DataFrame(data4)
df4.interpolate(method="linear",inplace=True)
df4
```

```python
# Aggregrate function

df4["Age"].mean()
df4["Age"].sum()
df4["Age"].max()
df4["Age"].min()
df4["Age"].std()
df4["Age"].count()
```

```python
# grouping row based on specific column

data5 ={
    "Name":["Ajay","Raj","Shiva","Aditi","Raja"],
    "Age":[20,21,22,20,22],
    "Salary":[5000,3000,4000,5000,4000]
}

df5=pd.DataFrame(data5)
df5

grouped=df5.groupby("Age")["Salary"].sum()
print(grouped)
```

```python
# merge table

df_merged=pd.merge(df4,df5,on="Name")
df_merged
```
