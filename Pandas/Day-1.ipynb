```python
import pandas as pd
```

```python
# create Series from list
country=['India','USA','China','Pakistan','Nepal']
print(country)
```

```python
country_series=pd.Series(country)
print(country_series)

# dtypes --> data type of value
# index   value
```

```python
# custom index

marks=[50,67,45,90]

subject=['hindi','english','math','science']

marks_series=pd.Series(marks,index=subject)
print(marks_series)

# index    value
```

```python
# write a name of value column
marks_series=pd.Series(marks,index=subject,name='student marks')
print(marks_series)
```

```python
# series from dictionary

marks={
    'maths':20,
    'english':60,
    'science':87
}

marks_series=pd.Series(marks,name='Shiva ka marks')
print(marks_series)

# key  act as index
# value act as value

# index    value
```

```python
# Series Attributes

# size   ---> return no. of value in series
print(marks_series.size)

# dtype  ---> return data type of value in series
print(marks_series.dtype)

# name  ---> return name of value column
print(marks_series.name)

# is_unique ---> return Ture if all items in value will unique otherwise return false
print(marks_series.is_unique)

# index   ---> return list of index
print(marks_series.index)

# values  ---> return numpy array of values
print(marks_series.values)
print(type(marks_series.values))
```
