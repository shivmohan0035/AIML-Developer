```python
# WAP find the sum of a 3 digit number entered by user
num=int(input("Enter 3 digit no.:"))
ans=num%10
num=num//10
ans+=num%10
num=num//10
ans+=num%10
print(ans)
```

```python
# if-else in Python
a=15
if a>15:
    print("greater than 15")
else: 
    print("less than 15")


# elif in Python
if a<15:
    print("less than 15")
elif a==15:
    print("equal to 15")
else: 
    print("greater than 15")

```

```python
# WAP to find min of 3 no.
a=int(input("Enter first no.:"))
b=int(input("Enter second no.:"))
c=int(input("Enter third no.:"))
if a<b and a<c:
    print("smallest is ",a)
elif b<c:
    print("smallest is ",b)
else: 
    print("smallest is ",c)
    
```
