```python
# break
for i in range(1,11):
  if i==5:
    break
  print(i,end=" ")
```

```python
# print all prime no between a range
lower=int(input("Enter lower:"))
upper=int(input("Enter upper:"))


for num in range(lower,upper+1):
  for i in range (2,num):
    if num%i==0 :
      break
  else:
    print(num,end=" ")
```

```python
# continue
for i in range(1,11):
    if i==5:
        continue
    print(i,end=" ")
        
```

```python
# string  ---> sequence of character in single or double quotation 
# it is immutable (not modify once created)

# Creating sting
s1="hello"
# ex--> "It's good" 
s2='hello'
# multiline string 
s3="""hello ji,
kaise ho!
thik ho!"""
s4=str("hello shiva!")
print(s1)
print(s2)
print(s3)
print(s4)
print(s1.upper())
print(s1)
```

```python
# operation on string

s1="Shiv"
s2="Mohan"
# Arithmatic operation ---> + , *

print(s1+" "+s2)
print(s1*5)

# Relational operation  ---> >, >=, < ,<=, ==, !=
print(s1>s2)   #compare using ASCII value
print(s1>=s2)
print(s1<s2)
print(s1<=s2)
print(s1!=s2)

# Logical operation ---> and , or, not
print("hello" and "shiv")
print("" and "shiv")
print("hello" or "shiv")
print("" or "shiv")
print(not "")
print(not " ")
```
